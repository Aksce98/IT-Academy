Com hem descobert que hi ha salaris en format text que es convertien en valors nulls, tornem a processar els salaris originals a la columna 'original' pero aquest cop no només fent la normalització dels valors numèrics, sino també dels valors textuals i ho guardem a 'Salari mensual' i eliminem la còpia a la columna 'original'.

```python
TEXT_SALARIES = {
    'mil vuit-cents': 1800,
    'dos mil': 2000,
    'tres mil': 3000,
}

def normalize_salary_with_text(series):
    def _parse(val):
        s = str(val).strip().lower()
        if s in TEXT_SALARIES:
            return TEXT_SALARIES[s]
        digits = re.sub(r'\D', '', s)
        return int(digits) if digits else None

    return pd.to_numeric(series.apply(_parse), errors='coerce')


df['Salari mensual'] = normalize_salary_with_text(df['original'])
df = df.drop(columns='original')
```

```python
def input_median(df, target, criteria):
  while len(criteria) > 0:
    df[target] = df[target].fillna(
        df.groupby(criteria)[target].transform('median')
    )
    del criteria[-1]

input_median(df, 'Salari mensual', ['Càrrec', 'Grup Professional', 'Ciutat', 'Gènere'])
input_median(df, 'Temperatura_mitjana_ciutat', ['País d\'origen', 'Ciutat'])
```

```python
df['children_adjustment'] = df['Nombre_fills'].apply(lambda x: 'no_children' if x == 0 else 'has_children').map(ps['children_adjustment'].dropna())
```

Afegim l'increment per edat al dataframe.

```python
from datetime import datetime

def rang_edat(fila):
    year = float(fila['Any de Naixement'])
    month = float(fila['Mes de Naixement'])
    day = float(fila['Dia de Naixement'])

    if pd.isna(year) or pd.isna(month) or pd.isna(day):
        return None

    today = datetime.today()
    age = today.year - int(year) - ((today.month, today.day) < (int(month), int(day)))

    if 18 <= age <= 29:
        return '18-29'
    elif 30 <= age <= 44:
        return '30-44'
    elif 45 <= age <= 59:
        return '45-59'
    elif age >= 60:
        return '60+'
    else:
        return None

df['age_adjustment'] = df.apply(rang_edat, axis=1).map(ps['age_adjustment'].dropna())
```

Afegim l'increment per cost de vida al dataframe.

```python
def cost_living(row, medium, high):
  c = row['Cost_vida_index']
  if high < c:
    return 'high'
  elif medium <= c < high:
    return 'medium'
  else:
    return 'low'

p33 = df['Cost_vida_index'].quantile(1/3)
p66 = df['Cost_vida_index'].quantile(2/3)
df['cost_of_living_adjustment'] = df.apply(cost_living, axis=1, args=(p33, p66)).map(ps['cost_of_living_adjustment'].dropna())
```

A continuació farem el mateix que cost_vida_index però aplicat a transport_public_score.

```python
def transport_score(row, medium, high):
  c = row['Transport_public_score']
  if c < medium:
    return 'low'
  elif medium <= c < high:
    return 'medium'
  else:
    return 'high'

p33 = df['Transport_public_score'].quantile(1/3)
p66 = df['Transport_public_score'].quantile(2/3)
df['transport_adjustment'] = df.apply(transport_score, axis=1, args=(p33, p66)).map(ps['transport_adjustment'].dropna())
```

A continuació, calcularem el nou salari aplicant un increment salarial amb una funció inventada ja que no se'ns proporciona cap fòrmula per aplicar.

```python
ps = pd.read_json('politica_salarial.json')
ps.fillna(' ')
```

```python
def calculate_new_salary(row, max_i, min):
  multiplicador = 1 + (row['base_increment'] +
                       row['children_adjustment'] +
                       row['age_adjustment'] +
                       row['cost_of_living_adjustment'] +
                       row['transport_adjustment']) / 100

  salary = round(row['Salari mensual'] * multiplicador, 2)
  if salary < min:
    return min
  elif row['Salari mensual'] * (1 + max_i/100) < salary:
    return salary * max_i
  else:
    return salary

max_increment = ps.loc['max_total_increment_percent', 'rules']
min_salary = ps.loc['min_salary_threshold', 'rules']
df['Salari mensual'] = df.apply(calculate_new_salary, axis=1, args=(max_increment, min_salary))
```
