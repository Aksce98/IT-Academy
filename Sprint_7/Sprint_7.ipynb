#Nivell 1 - Exercici 1: Selecció de Personal
La nostra funció principal serà main_seleccio().
Aquesta funció s'encarregarà de gestionar les dades i cridar a les funcions auxiliars.
La lògica de la funció serà la següent:
- Mentres no hi hagin 10 usuaris, obtindrà dades d'usuari i les validarà.
- Si les dades no són vàlides, les descarta i mostra un missatge d'error. Si són vàlides, obté la puntuació de l'usuari en funció de les dades i classifica la seva prioritat.
- Finalment afegeix les dades obtingudes i calculades a la nostra taula, i un cop tenim 10 usuaris a la nostra taula, mostrem l'informe final.

Les funcions auxiliars que utilitzarem seràn:
- obtenir_dades: s'encarregarà de demanar les dades a l'usuari. Per poder automatitzar les nostres proves, afegirem una variable global per poder utilitzar l'input d'exemple sense haver d'escriure les dades manualment
- validar_dades: verificarà que les dades tenen els valors permesos
- calcular_puntuacio: en funció de les dades, calcularà la puntuació de l'usuari
- classificar_candidat: classificarà l'usuari en funció de la seva puntuació
- mostrar_individual: mostrarà la puntuació de l'usuari
- mostrar_informe: ens mostrarà l'informe final amb totes les puntuacions

```python
def obtenir_dades(i):
  inputs_test = [
    "Laura Serra", "608112233", 6, 3, 3, 3,
    "Clara Gómez", "655221199", 3, 3, 3, 2,
    "Sergio López", "644112211", 3, 2, 2, 2,
    "Candidat Error", "611000000", 5, 7, 3, 2, # Català = 7 → NO VÀLID
    "Marta Ruiz", "699887766", 2, 3, 2, 1,
    "Marc Vidal", "612443322", 6, 3, 3, 3,
    "Júlia Romero", "677998877", 5, 3, 3, 2,
    "Pol Navarro", "699554433", 4, 3, 2, 2,
    "David Ferrer", "677665544", 2, 1, 1, 1,
    "Anna Paredes", "633221144", 1, 2, 1, 0,
    "Anibal Paredes", "644272727", 1, 2, 1, 0,
    "Núria Casals", "600112200", 8, 2, 3, 3 # 11è candidat
  ]
  i = i*6
  nom = inputs_test[i]
  telefon = inputs_test[i+1]
  experiencia = inputs_test[i+2]
  coneixement_catala = inputs_test[i+3]
  domini_excel = inputs_test[i+4]
  titol_academic = inputs_test[i+5]
  return nom, telefon, experiencia, coneixement_catala, domini_excel, titol_academic

def validar_dades(telefon, experiencia, coneixement_catala, domini_excel, titol_academic):
  if len(telefon) != 9:
    return False, "Telefon no té 9 digits"
  if experiencia < 0:
    return False, "La teva experiència no pot ser negativa"
  if not 0 <= coneixement_catala <= 3:
    return False, "El teu nivell de català ha d'estar entre 0 i 3"
  if not 0 <= domini_excel <= 3:
    return False, "El teu domini d'excel ha d'estar entre 0 i 3"
  if not 0 <= titol_academic <= 3:
    return False, "El teu titol acadèmic ha d'estar entre 0 i 3"
  return True, ""

def calcular_puntuacio(experiencia, coneixement_catala, domini_excel, titol_academic):
  return experiencia * 1.5 + coneixement_catala + domini_excel + titol_academic

def classificar_candidat(puntuacio):
  if 12 <= puntuacio:
    return "Prioritat Alta"
  if 8 <= puntuacio:
    return "Candidat adequat"
  return "No prioritari"

def mostrar_individual(nom, puntuacio, classificacio):
  print(f"{nom}, la teva puntuació és: {puntuacio}. Has estat classificat com a: {classificacio}")

def mostrar_informe(usuaris):
  usuaris.sort(key=lambda x: x[2], reverse = True)
  print("\n\nInforme final: ")
  for elem in usuaris:
    print(f"Nom: {elem[0]:<15} Telèfon: {elem[1]}    Puntuació: {elem[2]:<5}    Classificació: {elem[3]}")

def main_seleccio():
  usuaris = []
  usuari = 0
  while (len(usuaris) < 10):
    nom, telefon, experiencia, coneixement_catala, domini_excel, titol_academic = obtenir_dades(usuari)
    validesa_dades, error = validar_dades(telefon, experiencia, coneixement_catala, domini_excel, titol_academic)
    if validesa_dades:
      puntuacio = calcular_puntuacio(experiencia, coneixement_catala, domini_excel, titol_academic)
      classificacio = classificar_candidat(puntuacio)
      mostrar_individual(nom, puntuacio, classificacio)
      usuaris.append([nom, telefon, puntuacio, classificacio])
    else:
      print(f"Dades incorrectes. {error}")
    usuari += 1
  mostrar_informe(usuaris)

main_seleccio()
```
