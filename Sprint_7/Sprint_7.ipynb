#Nivell 1 - Exercici 1: Selecció de Personal
La nostra funció principal serà main_seleccio().
Aquesta funció s'encarregarà de gestionar les dades i cridar a les funcions auxiliars.
La lògica de la funció serà la següent:
- Mentres no hi hagin 10 usuaris, obtindrà dades d'usuari i les validarà.
- Si les dades no són vàlides, les descarta i mostra un missatge d'error. Si són vàlides, obté la puntuació de l'usuari en funció de les dades i classifica la seva prioritat.
- Finalment afegeix les dades obtingudes i calculades a la nostra taula, i un cop tenim 10 usuaris a la nostra taula, mostrem l'informe final.

Les funcions auxiliars que utilitzarem seràn:
- obtenir_dades: s'encarregarà de demanar les dades a l'usuari. Per poder automatitzar les nostres proves, afegirem una variable global per poder utilitzar l'input d'exemple sense haver d'escriure les dades manualment
- validar_dades: verificarà que les dades tenen els valors permesos
- calcular_puntuacio: en funció de les dades, calcularà la puntuació de l'usuari
- classificar_candidat: classificarà l'usuari en funció de la seva puntuació
- mostrar_individual: mostrarà la puntuació de l'usuari
- mostrar_informe: ens mostrarà l'informe final amb totes les puntuacions

```python
def obtenir_dades(i):
  inputs_test = [
    "Laura Serra", "608112233", 6, 3, 3, 3,
    "Clara Gómez", "655221199", 3, 3, 3, 2,
    "Sergio López", "644112211", 3, 2, 2, 2,
    "Candidat Error", "611000000", 5, 7, 3, 2, # Català = 7 → NO VÀLID
    "Marta Ruiz", "699887766", 2, 3, 2, 1,
    "Marc Vidal", "612443322", 6, 3, 3, 3,
    "Júlia Romero", "677998877", 5, 3, 3, 2,
    "Pol Navarro", "699554433", 4, 3, 2, 2,
    "David Ferrer", "677665544", 2, 1, 1, 1,
    "Anna Paredes", "633221144", 1, 2, 1, 0,
    "Anibal Paredes", "644272727", 1, 2, 1, 0,
    "Núria Casals", "600112200", 8, 2, 3, 3 # 11è candidat
  ]
  i = i*6
  nom = inputs_test[i]
  telefon = inputs_test[i+1]
  experiencia = inputs_test[i+2]
  coneixement_catala = inputs_test[i+3]
  domini_excel = inputs_test[i+4]
  titol_academic = inputs_test[i+5]
  return nom, telefon, experiencia, coneixement_catala, domini_excel, titol_academic

def validar_dades(telefon, experiencia, coneixement_catala, domini_excel, titol_academic):
  if len(telefon) != 9:
    return False, "Un telèfon d'Espanya ha de tenir 9 digits"
  if experiencia < 0:
    return False, "La teva experiència no pot ser negativa"
  if not 0 <= coneixement_catala <= 3:
    return False, "El teu nivell de català ha d'estar entre 0 i 3"
  if not 0 <= domini_excel <= 3:
    return False, "El teu domini d'excel ha d'estar entre 0 i 3"
  if not 0 <= titol_academic <= 3:
    return False, "El teu titol acadèmic ha d'estar entre 0 i 3"
  return True, ""

def calcular_puntuacio(experiencia, coneixement_catala, domini_excel, titol_academic):
  return experiencia * 1.5 + coneixement_catala + domini_excel + titol_academic

def classificar_candidat(puntuacio):
  if 12 <= puntuacio:
    return "Prioritat Alta"
  if 8 <= puntuacio:
    return "Candidat adequat"
  return "No prioritari"

def mostrar_individual(nom, puntuacio, classificacio):
  print(f"{nom}, la teva puntuació és: {puntuacio}. Has estat classificat com a: {classificacio}")

def mostrar_informe(usuaris):
  usuaris.sort(key=lambda x: x[2], reverse = True)
  print("\n\nInforme final: ")
  for elem in usuaris:
    print(f"Nom: {elem[0]:<15} Telèfon: {elem[1]}    Puntuació: {elem[2]:<5}    Classificació: {elem[3]}")

def main_seleccio():
  usuaris = []
  usuari = 0
  while (len(usuaris) < 10):
    nom, telefon, experiencia, coneixement_catala, domini_excel, titol_academic = obtenir_dades(usuari)
    validesa_dades, error = validar_dades(telefon, experiencia, coneixement_catala, domini_excel, titol_academic)
    if validesa_dades:
      puntuacio = calcular_puntuacio(experiencia, coneixement_catala, domini_excel, titol_academic)
      classificacio = classificar_candidat(puntuacio)
      mostrar_individual(nom, puntuacio, classificacio)
      usuaris.append([nom, telefon, puntuacio, classificacio])
    else:
      print(f"Dades incorrectes. {error}")
    usuari += 1
  mostrar_informe(usuaris)

main_seleccio()
```

- obtenir_segment_i_factor(): rebrà un client i retornarà el seu segment i el factor_codi
- obtenir_factor_edat(): rebrà un client i retornarà calculat el factor_edat
- calcular_risc(): rebrà el factor d'edat i de codi i calcularà el risc utilitzant la fòrmula risc_final = factor codi * factor edat
- mostrar_diccionari(): rebrà el diccionari de clients, l'ordenarà per ordre descendent de risc final y el mostrará per pantalla.
- mostrar_respostes(): calcularà i mostrarà les respostes de les 4 preguntes proposades a l'enunciat



```python
def obtenir_segment_i_factor(client):
  segment = ["Navega però no afegeix productes",
             "Afegeix productes però no arriba al checkout",
             "Arriba al checkout però no paga",
             "Cistella de compra abandonat fa menys de 24 h",
             "Cistella de compra abandonat fa més de 24 h"]
  factor_codi = [0.1,0.25,0.50,0.75,0.90]
  index = client["codi_comportament"]
  return segment[index], factor_codi[index]

def obtenir_factor_edat(client):
  factor_edat = {"18-25" : 1.20,
                 "26-40" : 1.00,
                 "41-60" : 0.85,
                 "60+" : 0.70}
  return factor_edat[client["grup_edat"]]

def calcular_risc(factor_codi, factor_edat):
  return factor_codi * factor_edat

def mostrar_diccionari(diccionari):
  diccionari = dict(sorted(diccionari.items(), key=lambda x: x[1]['risc_final'], reverse=True))
  for value, key in diccionari.items():
    print(value)
    print(key)

def mostrar_respostes(diccionari):
  print("\n1. Quins clients s'han de contactar primer en una campanya de retargeting?")
  print("Aquells que tenen un risc final més alt.")

  print("\n2. Quines diferències hi ha entre clients amb el mateix comportament però d'edats diferents? (Per exemple: un client jove que abandona la cistella de compra fa 24 h té el mateix risc que un de més edat?)")
  print("Si no variem el codi de comportament, quan augmenta l'edat, disminueix el risc.")

  print("\n3. Quins clients presenten tan poc interès que no val la pena invertir-hi pressupost?")
  print("Els de major edat amb un codi de comportant més baix, com pot ser l'Eva Martín o el Luís García.")
  return

def processar_clients(clients):
  diccionari = {}
  for client in clients:
    segment, factor_codi = obtenir_segment_i_factor(client)
    factor_edat = obtenir_factor_edat(client)
    risc_final = calcular_risc(factor_codi, factor_edat)
    client["segment"] = segment
    client["factor_codi"] = factor_codi
    client["factor_edat"] = factor_edat
    client["risc_final"] = risc_final
    id_client = int(client.pop("id_client"))
    diccionari[id_client] = client
  return diccionari

def main_marketing(clients):
  diccionari = processar_clients(clients)
  mostrar_diccionari(diccionari)
  mostrar_respostes(diccionari)


clients = [
    {"id_client": 1, "nom": "Laura Serra", "edat": 23, "grup_edat": "18-25", "codi_comportament": 4},
    {"id_client": 2, "nom": "Marc Vidal", "edat": 35, "grup_edat": "26-40", "codi_comportament": 3},
    {"id_client": 3, "nom": "Ana López", "edat": 29, "grup_edat": "26-40", "codi_comportament": 2},
    {"id_client": 4, "nom": "Joan Riera", "edat": 19, "grup_edat": "18-25", "codi_comportament": 2},
    {"id_client": 5, "nom": "Mónica Pérez", "edat": 47, "grup_edat": "41-60", "codi_comportament": 4},
    {"id_client": 6, "nom": "Luis García", "edat": 52, "grup_edat": "41-60", "codi_comportament": 1},
    {"id_client": 7, "nom": "Pilar Sánchez", "edat": 61, "grup_edat": "60+", "codi_comportament": 3},
    {"id_client": 8, "nom": "Eva Martín", "edat": 38, "grup_edat": "26-40", "codi_comportament": 0},
    {"id_client": 9, "nom": "Diego Romero", "edat": 24, "grup_edat": "18-25", "codi_comportament": 1},
    {"id_client": 10, "nom": "Núria Costa", "edat": 33, "grup_edat": "26-40", "codi_comportament": 4}
]

main_marketing(clients)
```

#Nivell 2 - Exercici 1: Gestió de risc i priorització de pacients
La funció principal d'aquest exercici serà processar_pacients(), que s'encarregarà de rebre un llistat de pacients i de nivells de risc, i mostrarà per pantalla aquest llistat de pacients però ordenat segons el nivell de risc de manera descendent.
Per poder realitzar això, la funció processar_clients utilitzarà diverses funcions auxiliars:
- separar_pacient(): rep un string d'un pacient, i separa el nom, l'edat i els símptomes.
- normalitzar_simptomes(): rep un string de símptomes, separa aquest string per comes, converteix totes les lletres a minúscules i elimina cualsevol símbol que no sigui una lletra, després converteix aquest nou string en el símptoma correcte i finalment retorna una llista de simptomes. Per exemple, convertiría "dolor CABeZA" a "dolorcabeza" a "dolor_cabeza".
- sanitize(): rep un string, elimina tot el que no siguin lletres a-z o A-Z i ho converteix a minúscules.
- calcular_risc(): rep un llistat de símptomes i retorna el risc del pacient.
- mostrar_pacients(): rep un llistat de pacients, l'ordena per risc en ordre descendent i ho mostra per pantalla.

```python
import re
import pandas as pd

def separar_pacient(pacient):
  elements = pacient.split("|")
  return elements[0], elements[1], elements[2]

def sanitize(simptoma):
    return re.sub(r'[^a-zA-Z]', '', simptoma.lower())

def normalitzar_simptomes(simptomes, nivells_risc):
  risc = {} #Creem un diccionari per transformar simptomes sanitizats en el simptoma correcte. Per exemple, "dolorcabeza" a "dolor_cabeza"
  for r in nivells_risc:
    risc[sanitize(r)] = r

  simptomes = simptomes.split(",")
  simptomes_normalitzats = []
  for s in simptomes:
    simptomes_normalitzats.append(risc[sanitize(s)])
  return simptomes_normalitzats

def calcular_risc(simptomes, nivells_risc):
  risc = 0
  for s in simptomes:
    risc += nivells_risc[s]
  return risc

def mostrar_pacients(df):
  df = df.sort_values(by='Risc', ascending=False)
  print(df)

def processar_pacients(pacients, nivells_risc):
  df = pd.DataFrame(columns=['Nom', 'Edat', 'Simptomes', 'Risc'])
  for pacient in pacients:
    nom, edat, simptomes = separar_pacient(pacient)
    simptomes = normalitzar_simptomes(simptomes, nivells_risc)
    risc = calcular_risc(simptomes, nivells_risc)
    df.loc[len(df)] = [nom, edat, simptomes, risc]
  mostrar_pacients(df)

pacients = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
]

nivells_risc = {
    "fiebre_alta": 3,
    "dificultad_respirar": 5,
    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,
    "dolor_cabeza": 1
}

processar_pacients(pacients, nivells_risc)
```

# Nivell 3 - Exercici 1
En aquest exercici tenim 2 opcions per crear el nostre string:
- Generar el string de manera aleatòria i verificar que compleix les condicions demanades.
- Afegir 1 caràcter de cada condició i omplir la resta de N-condicions caràcters de manera aleatòria.

La primera opció és més senzilla d'implementar, però en el pitjor dels casos, que seria un string de 4 caràcters amb els 4 tipus, tindriem una probabilitat de generar l'string correctament un 6.65% de les vegades. Això be donat per que tenim 4! x 26 x 26 x 10 x 32 combinacions correctes sobre un total de (26+26+10+32)ˆ4 combinacions.
En un cas qualsevol amb un string de major mida, la probabilitat de generar-ho correctament a la primera és molt elevada ja que tenim molts intents per col·locar correctament els tipus demanats.

La segona opció sempre generarà una contrasenya vàlida a la primera, però no és estricament millor ja que tindrà un bias a generar strings amb exactament 1 element de cada tipus.

Per tant, implementarem la primera opció amb un senzill bucle que generarà el string i validarà que compleixi les condicions.
Utilitzarem 3 funcions auxiliars:
- Una funció per crear el 'pool' de caràcters
- Una funció per generar la contrasenya
- Una funció per validar la contrasenya

```python
import string
import numpy as np

def caracters_possibles(majuscules, minuscules, numeros, signes):
  pool = []
  if majuscules:
    pool += string.ascii_uppercase
  if minuscules:
    pool += string.ascii_lowercase
  if numeros:
    pool += string.digits
  if signes:
    pool += string.punctuation
  return pool

def crear_string(mida, caracters):
  rng = np.random.default_rng()
  return rng.choice(caracters, size = mida)

def comprobar_validesa(contrasenya, majuscules, minuscules, numeros, signes):
  validesa = True
  if majuscules and not (set(string.ascii_uppercase) & set(contrasenya)):
    return False
  if minuscules and not (set(string.ascii_lowercase) & set(contrasenya)):
    return False
  if numeros and not (set(string.digits) & set(contrasenya)):
    return False
  if signes and not (set(string.punctuation) & set(contrasenya)):
    return False
  return True

def crear_contrasenya(mida, majuscules = True, minuscules = True, numeros = True, signes = True):
  #La mida de la contrasenya mai pot ser inferior a la quantitat de condicions que tenim
  assert majuscules + minuscules + numeros + signes <= mida, 'Fes la contrasenya més llarga o redueix els requisits!'

  contrasenya = []
  valid = False
  while not valid:
    caracters = caracters_possibles(majuscules, minuscules, numeros, signes)
    contrasenya = crear_string(mida, caracters)
    valid = comprobar_validesa(contrasenya, majuscules, minuscules, numeros, signes)
  contrasenya = ''.join(contrasenya)
  return contrasenya

print(crear_contrasenya(4))
print(crear_contrasenya(12))
print(crear_contrasenya(12, majuscules = False, minuscules = False))
print(crear_contrasenya(20, numeros = False, majuscules = False))
print(crear_contrasenya(12, majuscules = False, minuscules = False, signes = False))
print(crear_contrasenya(3, numeros = False))
print(crear_contrasenya(3))
```

#Nivell 3 - Exercici 3
En aquest exercici utilitzarem una funció principal processar_dades_esportives() que rebrà un string que indiqui la ruta de l'històric de partits.
Convertirem aquest històric de partits en una llista on cada element és un partit individual i utilitzant aquests partits crearem el nostre dataframe.
A cada partit, verificarem si els equips ja estàn presents al dataframe, si no, els inicialitzem amb tots els valors a 0.
Si ja estàn presents, simplement hem d'actualitzar les dades dels gols, victoria/empat/derrota i afegir els punts corresponents.

```python
import pandas as pd
import re
import time

def llegir_fitxer(ruta):
  with open(ruta, 'r', encoding='utf-8') as f:
    partits = f.read().splitlines()
    return partits

def processar_partit(partit):
  #Utilitzem regex ja que un simple split no podria diferencia l'espai
  #entre equip i resultat o simplement un espai entre el nom de l'equip
  partit = result = [x.strip() for x in re.split(r'(\d+-\d+)', partit, maxsplit=1)]
  local = partit[0]
  resultat = partit[1]
  visitant = partit[2]
  resultat = resultat.split("-")
  gols_local = resultat[0]
  gols_visitant = resultat[1]
  return local, int(gols_local), visitant, int(gols_visitant)

def actualitzar_dataframe(df, equip, gols_favor, gols_contra):
  mask = df['equip'] == equip
  df.loc[mask, 'gols_favor'] = df.loc[mask, 'gols_favor'] + gols_favor
  df.loc[mask, 'gols_contra'] = df.loc[mask, 'gols_contra'] + gols_contra
  if gols_favor > gols_contra:
    df.loc[mask, 'guanyats'] += 1
    df.loc[mask, 'punts'] += 3
  elif gols_favor == gols_contra:
    df.loc[mask, 'empatats'] += 1
    df.loc[mask, 'punts'] += 1
  else:
    df.loc[mask, 'perduts'] += 1
  return df

def inicialitzar_equip(df, equip):
  df.loc[len(df)] = {'equip':equip,'gols_favor':0, 'gols_contra':0, 'guanyats':0,'empatats':0,'perduts':0,'punts':0}
  return df

def equip_mes_golejador(df):
  df = df.sort_values(by='gols_favor', ascending = False)
  print(f'Equip més golejador: {df['equip'].iloc[0]}')

def equip_mes_golejat(df):
  df = df.sort_values(by='gols_contra', ascending = False)
  print(f'Equip més golejat: {df['equip'].iloc[0]}')

def mostrar_classificacio(df):
  df = df.sort_values(by='punts', ascending = False)
  equips = df['equip'].tolist()
  print('\nClassificació:')
  for i in range(0, len(equips)):
    print(f'{i}. {equips[i]}')

def processar_dades_esportives(ruta):
  partits = llegir_fitxer(ruta)
  df = pd.DataFrame(columns=['equip','gols_favor','gols_contra','guanyats','empatats','perduts','punts'])
  for partit in partits:
    local, gols_local, visitant, gols_visitant = processar_partit(partit)
    if not df['equip'].isin([local]).any():
      df = inicialitzar_equip(df, local)
    if not df['equip'].isin([visitant]).any():
      df = inicialitzar_equip(df, visitant)
    df = actualitzar_dataframe(df, local, gols_local, gols_visitant)
    df = actualitzar_dataframe(df, visitant, gols_visitant, gols_local)
  print(df)
  equip_mes_golejador(df)
  equip_mes_golejat(df)
  mostrar_classificacio(df)

processar_dades_esportives("historic partits.txt")
```
