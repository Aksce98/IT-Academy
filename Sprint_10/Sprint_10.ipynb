# Nivell 1

```python
import requests
import pandas as pd
```

## 1. Consulta l’API pública JSONPlaceholder utilitzant el mètode GET per obtenir:

```python
link = 'https://jsonplaceholder.typicode.com/'
```

### Llista de publicacions (/posts)

```python
r_posts = requests.get(link+'posts')
df_posts = pd.DataFrame.from_dict(r_posts.json())
df_posts.head(5)
```

### Llista d’usuaris (/users)

```python
r_users = requests.get(link+'users')
df_users = pd.DataFrame.from_dict(r_users.json())
df_users.head(5)
```

### Llista de tasques (/todos)

```python
r_todos = requests.get(link+'todos')
df_todos = pd.DataFrame.from_dict(r_todos.json())
df_todos.head(5)
```

## 2. Mostra per pantalla:

### La quantitat total de cada recurs.

```python
print("Total de posts:", len(df_posts))
print("Total d'usuaris:", len(df_users))
print("Total de tasques:", len(df_todos))
```

### El codi d'estat de cada petició

```python
print(f'El codi de estat de posts és {r_posts.status_code}')
print(f'El codi de estat de users és {r_users.status_code}')
print(f'El codi de estat de todos és {r_todos.status_code}')
```

Els endpoints disponibles a aquesta API són:

GET /api/holidays lists every country; add a 2-letter code for that country’s 2026 public holidays. CORS-open, no key required.
- Automatic bridge-day planner.
GET /api/long-weekends/{cc} — every 3+ day break in 2026 (weekends + public holidays), with the bridge days worth booking off.
- Live local time, DST-correct.
GET /api/clock — current time for caldays city codes (?cities=) and/or any IANA zone (?tz=), computed from the IANA tz database.

Els filtres opcionals que ofereix són:
- Country Code: Ens permet filtrar les dades per pais a l'endpoint de holidays i long-weekends.
- cities o tz: ens permet filtrar per ciutat a l'endpoint de clock.

# Nivell 3

## 1. Utilitza l’API d’Open Data BCN per:

Cercar un dataset del teu interès mitjançant package_search i package_show

```python
base_url = 'https://opendata-ajuntament.barcelona.cat/data/api/'
params = {'limit' : 1000}

r_bcn = requests.get(base_url + 'action/current_package_list_with_resources', params=params)
r_bcn.status_code
```

```python
df = pd.DataFrame.from_dict(r_bcn.json()['result'])
df
```

Com que no m'interesa cap dataset en particular, farem la selecció de manera aleatoria.

## 2. Dels resultats obtinguts, selecciona’n un que tingui recursos disponibles en CSV o JSON.

Fem un explode de 'resources' per obtenir les metadades de cada dataset.

```python
resources = pd.json_normalize(df.explode('resources')['resources'])
```

Escollim un dataset aleatori entre aquells que tenen format CSV.

```python
id = resources[resources['format'] == 'CSV'].sample(1, random_state = 1)['id'].values[0]
print(id)
```

## 3. Amb el resource_id del dataset seleccionat, realitza una consulta mitjançant datastore_search per:

Construïm la url i fem el request amb limit 100 per obtenir 100 resultats.

```python
params = {'resource_id':id, 'limit':100}

r = requests.get(base_url + 'action/datastore_search', params)
r.status_code
```

## 4. Converteix els resultats a un DataFrame.

```python
pkg_id = resources.loc[resources['id'] == id, 'package_id'].values[0]
title = df.loc[df['id'] == pkg_id, 'title_translated'].values[0]['ca']
print(title)
```

```python
resultat = pd.DataFrame.from_dict(r.json()['result']['records'])
resultat
```

## 5. Desa el DataFrame en un fitxer .csv.

```python
resultat.to_csv(str(title)+'.csv', index=False)
```
